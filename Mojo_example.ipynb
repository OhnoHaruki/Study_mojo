## Mojoファイルの作成

```
%%file print_mojo.mojo
from sys import argv

fn print_mojo(number: UInt8):
    for _ in range(number):
        print("🔥", end="")

fn main():
    var args = argv()
    var n: UInt8

    try:
        n = atol(args[1])
    except:
        n = 1

    print_mojo(n)
```

## ビルド

```
!mojo build print_mojo.mojo -o print_mojo
```

## 実行
- 上がビルドしたものを実行する。下はそのまま即時実行
